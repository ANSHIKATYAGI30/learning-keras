```python
from tensorflow.keras import Model
```

```python
input_dims = X_train.shape[1]
output_dims = 10

#input layer
input_layer = tf.keras.Input(Shape = (input_dims,), name = "input_layer")

#first hidden layer
hidden_layer1 = Dense(512, activation = "sigmoid", name = "Hidden_Layer1")(input_layer)
hidden_layer2 = Dense(128, activation = "sigmoid", name = "Hidden_Layer2")(hidden_layer1)

#output layer
output_layer = Dense(output_dims, activation = "softmax", name = "Output_Layer")(hidden_layer2)

model = Model(inputs = input_layer, outputs = output_layer)
```

```python
from tensorflow.keras.utils import plot_model
plot_model(model)
```

```python
#compiling the model
```

```python
model.compile(loss = "sparse_categorical_crossentropy", optimizer = "adam", metrics = ["accuracy"])
```

```python
#training the neural network
#define the batch size and no. of epochs
batch_size = 126 #should always be in the form of 2^n
epochs = 20

#fit a neural network or train a neural network
history = model.fit(X_train, y_train, batch_size = batch_size, epochs = epochs, verbose = 1, validation_data = (X_test, y_test))
```

```python
#evaluating a trained network
score = model.evaluate(x = X_test, y = y_test, verbose = 0)
print("Test loss: ", score[0])
print("Test accuracy: ", score[1])
```
